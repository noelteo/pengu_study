```python
# 2.3.7 INT/FLOAT 
# Arithmetic operators
a // b # floor division -> rounded down value
a % b # modulo (remainder)
a ** b # exponentiation

# Rounding
round(x) # round normally
math.floor(x) # round down
math.ceil(x) # round up
math.trunc(x) # round towards zero

# Square roots
math.sqrt(x)

# generation of ranges
range(x) # 0 to x - 1
range(x,y) # x to y-1
range(x,y,z) # x to y-1, increment by z

# generation of random ints/floats
random.randint(x,y) # returns a random int between x and y (inclusive)
random.random() # returns a random float where 0 <= float < 1

# conversion of strings to floats/ints
float(x)
int(x)
```

```python
# 2.3.8 STRINGS
# escape codes
\ # ignore end of line
\n # newline
\t # tab

# concatenation and repetition
a + b # concatenation (both must be str)
a * b # repetition (one str and one int)

# indexing and slicing
str[i] # returns i-th position of str (if i is negative count from back, -1 is the last)
str[a:b] # returns str from index a to b-1
str[a:b:c] # returns str from index a to b-1, incrementing by c.

# conversion to and from ASCII
chr(int) # returns str containing character of ASCII code
ord(str) # returns int of ASCII code of character

# Conversion to uppercase/lowercase
str.upper()
str.lower()

# Testing of whether characters are letters only, lowercase letters only, uppercase letters only, digits only, spaces only and or alphanumeric
# ALL RETURNS BOOL
str.isalpha() #True if every character is letters
str.islower() #True if every letter is uppercase
str.isupper() #True if every letter is lowercase
str.isdigit() #True if every character is digit
str.isspace() #True if every character is space
str.isalnum() #True if every character is letter or digit

# substring in string? MEMBERSHIP
substr in str
substr not in str

# Testing of whether the string starts and/or ends with a substring
str.startswith(x)
str.endswith(x)

# searching for location of substring
str.find(x) # x is the substring, returns the index of first character of x
str.find(x,a,b) # search from a to b-1 only.

# Splitting strings into list
str.split() # split based on spaces
str.split(x) # split based on x (delimiter)

# calculation of lenggthh
len(str)

# converting to strings
str(x)

# output formatting
f"... {var} ..."
str.format() ->  " {} {} ".format(x,y)
```
